## Freight Cost Prediction

- **Overview:** This analysis focuses on predicting freight cost using vendor invoice data like quantity and price.
- **Purpose:** It supports cost estimation, invoice validation, and identification of unusual freight charges.
- **Objective:** The goal is to develop a reliable model for accurate and efficient freight cost prediction to improve cost forecasting, budgeting, and vendor negotiation.

```python
vendor_df = pd.read_sql_query("select * from vendor_invoice",conn)
vendor_df.head()
```

```python
# Relationship between 

sns.heatmap(vendor_df[["Quantity", "Freight", "Dollars"]].corr(), annot=True, cmap="viridis")
plt.show()
```

```python
plt.scatter(vendor_df["Quantity"], vendor_df["Freight"], color="blue")
plt.scatter(vendor_df["Dollars"], vendor_df["Freight"], color="orange")
plt.legend(["Quantity", "Dollars"])
plt.ylabel("Freight Cost")
plt.show()
```

```python
vendor_df['freight_per_unit'] = vendor_df["Freight"]/vendor_df["Quantity"]
vendor_df.head()
```

```python
low_quantity = vendor_df["Quantity"].quantile(0.25)
high_quantity = vendor_df["Quantity"].quantile(0.75)

print(f"Low Quantity {low_quantity}")
print(f"High Quantity {high_quantity}")
```

```python
vendor_df.loc[vendor_df["Quantity"]<low_quantity, 'freight_per_unit'].mean()
```

```python
vendor_df.loc[vendor_df["Quantity"]>high_quantity, 'freight_per_unit'].mean()

# Higher the quantity, lower the freight cost
```

```python
X = vendor_df[['Dollars']]
y = vendor_df['Freight']
```

```python
vendor_df.describe().round()

# If you look at the 75% and the max value, there is a huge difference. This means there are outliers in the data
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=101)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
model1 = LinearRegression()
model1.fit(X_train, y_train)

model2 = DecisionTreeRegressor(max_depth=4,random_state=101)
model2.fit(X_train, y_train)

model3 = RandomForestRegressor(max_depth=3,random_state=101)
model3.fit(X_train, y_train)
```

```python
def evaluate_models(model, X_test, y_test, model_name):
    y_pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)*100

    print(f"\n {model_name} Performance: ")
    print(f" MAE Performance: {mae:.2f}")
    print(f" RMSE Performance: {rmse:.2f}")
    print(f" R2 Score Performance: {r2:.2f}%")
```

```python
evaluate_models(model1, X_test, y_test, 'Linear Regression')
evaluate_models(model2, X_test, y_test, 'Descision Tree Regression')
evaluate_models(model3, X_test, y_test, 'Random Forest Regression')
```
